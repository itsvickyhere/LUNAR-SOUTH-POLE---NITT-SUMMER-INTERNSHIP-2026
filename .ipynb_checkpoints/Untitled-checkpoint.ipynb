```python
"""
=============================================================================
STEP 2 — CANDIDATE REGION DEFINITION
Method: Watershed Terrain Segmentation on LOLA Slope
(Object-Based Image Analysis / Land-Surface Segmentation)
=============================================================================

MATHEMATICAL BASIS
------------------
1. slope(x,y) = arctan( sqrt( (dz/dx)² + (dz/dy)² ) )   [degrees]
   via Sobel finite-difference gradient on LOLA elevation

2. roughness(x,y) = std(elevation in N×N kernel)
   local terrain variability at each pixel

3. Watershed on slope:
   a) gradient_img = sobel(slope)      → high at slope breaks (crater rims etc.)
   b) markers = local minima of slope  → flattest spots = seeds per terrain unit
   c) watershed(gradient_img, markers) → floods from seeds, stops at gradient peaks
   → each region is bounded by REAL terrain discontinuities, not arbitrary lines

4. Multi-resolution sampling:
   - High-res (LOLA 5m/25m): pixel-wise stats inside region (guaranteed homogeneous)
   - Mid-res (DFSAR 25m, Mini-RF 59m): centroid-based nearest-pixel sample
   - Coarse (M3 280m, GRAIL 10km): nearest-pixel lookup = regional context constant

REFERENCE
---------
Drăguţ & Blaschke (2006) "Automated classification of landform elements using
  object-based image analysis." Geomorphology 81(3-4):330-344.
Drăguţ & Eisank (2011) "Object representations at multiple scales from DEMs."
  Geomorphology 129(3-4):183-189.

INSTALL (run once)
------------------
pip install rasterio scikit-image scipy numpy pandas matplotlib
=============================================================================
"""

# ── INSTALL CHECK ──────────────────────────────────────────────────────────
import subprocess, sys

def ensure(pkg, imp=None):
    try: __import__(imp or pkg)
    except ImportError:
        print(f"Installing {pkg}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg,
                               "--break-system-packages", "-q"])

for pkg, imp in [("rasterio",None),("scikit-image","skimage"),
                 ("scipy",None),("numpy",None),("pandas",None),("matplotlib",None)]:
    ensure(pkg, imp)

# ── IMPORTS ────────────────────────────────────────────────────────────────
import os, warnings
import numpy as np
import pandas as pd
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import rasterio
from rasterio.transform import rowcol, xy as rc_xy
from scipy.ndimage import generic_filter, label as sp_label, minimum_filter
from skimage.filters import sobel
from skimage.segmentation import watershed
from skimage.measure import regionprops, label as sk_label

warnings.filterwarnings("ignore")

# ══════════════════════════════════════════════════════════════════════════
# 0. CONFIGURATION
# ══════════════════════════════════════════════════════════════════════════

BASE = r"D:\INTERN - LUNAR SOUTH POLE\Intern_3_granularity_science_priority\raster_layers"
OUT  = r"D:\INTERN - LUNAR SOUTH POLE\Intern_3_granularity_science_priority\outputs"
os.makedirs(OUT, exist_ok=True)

PATHS = {
    "lola_25m":    os.path.join(BASE, "lola_downsampled_25m.tif"),
    "shadow":      os.path.join(BASE, "lroc_shadow_mask_final.tif"),
    "dfsar_cpr":   os.path.join(BASE, "dfsar_cpr_mosaic_circle.tif"),
    "dfsar_cov":   os.path.join(BASE, "dfsar_cpr_mosaic_coverage.tif"),
    "minirf_s1":   os.path.join(BASE, "minirf_s1_pju_mosaic_circle.tif"),
    "minirf_cov":  os.path.join(BASE, "minirf_s1_pju_mosaic_coverage.tif"),
    "m3":          os.path.join(BASE, "m3_mosaic_group2_july13_14_allbands_280m.tif"),
    "grail_anom":  os.path.join(BASE, "gravil_anom_l900_native_10km_circular.tif"),
    "grail_boug":  os.path.join(BASE, "gravil_boug_l900_native_10km_circular.tif"),
}

# Tunable parameters
SEED_WINDOW_PX   = 15      # search window for local slope minima (pixels)
SEED_MAX_SLOPE   = 20.0    # only seed from pixels with slope < this (degrees)
ROUGHNESS_KERNEL = 7       # NxN kernel for roughness calculation

print("=" * 65)
print("STEP 2 — CANDIDATE REGION DEFINITION")
print("Method: Watershed Terrain Segmentation (OBIA / Land-Surface Seg)")
print("=" * 65)

# ══════════════════════════════════════════════════════════════════════════
# 1. LOAD LOLA 25m
# ══════════════════════════════════════════════════════════════════════════

print("\n[1/6] Loading LOLA 25m elevation...")
with rasterio.open(PATHS["lola_25m"]) as src:
    elev_raw  = src.read(1).astype(np.float64)
    nodata    = src.nodata
    transform = src.transform
    crs       = src.crs
    res_x     = abs(transform.a)
    res_y     = abs(transform.e)
    H, W      = elev_raw.shape

valid_mask = (elev_raw != nodata) & np.isfinite(elev_raw) if nodata is not None \
             else np.isfinite(elev_raw)
elev       = np.where(valid_mask, elev_raw, np.nan)
elev_fill  = np.nan_to_num(elev, nan=float(np.nanmean(elev_raw[valid_mask])))

print(f"   {H}×{W}  @ {res_x:.1f}m/px  |  valid: {valid_mask.sum():,}/{elev_raw.size:,}")

# ══════════════════════════════════════════════════════════════════════════
# 2. SLOPE + ROUGHNESS
# ══════════════════════════════════════════════════════════════════════════

print("\n[2/6] Computing slope and roughness...")
dy_px, dx_px = np.gradient(elev_fill, res_y, res_x)
slope_deg  = np.degrees(np.arctan(np.sqrt(dx_px**2 + dy_px**2)))
slope_deg  = np.where(valid_mask, slope_deg, np.nan)

roughness  = generic_filter(elev_fill, np.std, size=ROUGHNESS_KERNEL, mode="nearest")
roughness  = np.where(valid_mask, roughness, np.nan)

print(f"   Slope: {np.nanmin(slope_deg):.2f}° – {np.nanmax(slope_deg):.2f}°  "
      f"(mean {np.nanmean(slope_deg):.2f}°)")

# ══════════════════════════════════════════════════════════════════════════
# 3. WATERSHED SEGMENTATION
# ══════════════════════════════════════════════════════════════════════════

print(f"\n[3/6] Watershed segmentation  (seed window={SEED_WINDOW_PX}px, "
      f"max seed slope={SEED_MAX_SLOPE}°)...")

slope_fill = np.nan_to_num(slope_deg, nan=float(np.nanmax(slope_deg[valid_mask])))

# Gradient image: high at terrain boundaries (crater rims, slope breaks)
gradient_img = sobel(slope_fill)

# Seeds: local minima of slope = flattest spots in each terrain unit
local_min  = slope_fill == minimum_filter(slope_fill, size=SEED_WINDOW_PX)
local_min &= (slope_deg < SEED_MAX_SLOPE) & valid_mask
markers, n_seeds = sp_label(local_min)
print(f"   Seeds (local slope minima): {n_seeds}")

# Flood from seeds outward, stop at gradient ridges
region_labels = watershed(gradient_img, markers, mask=valid_mask)
n_regions     = region_labels.max()
print(f"   Regions produced: {n_regions}")

# ══════════════════════════════════════════════════════════════════════════
# 4. LOAD SECONDARY DATASETS
# ══════════════════════════════════════════════════════════════════════════

print("\n[4/6] Loading secondary datasets...")

def load_band(path, band=1):
    with rasterio.open(path) as s:
        d = s.read(band).astype(np.float64)
        nd, tf = s.nodata, s.transform
    if nd is not None:
        d = np.where(d == nd, np.nan, d)
    return d, tf

def geo_sample(data, tf, gx, gy):
    r, c = rowcol(tf, gx, gy)
    r, c = int(r), int(c)
    if 0 <= r < data.shape[0] and 0 <= c < data.shape[1]:
        v = data[r, c]
        return float(v) if np.isfinite(v) else np.nan
    return np.nan

# Shadow (5m) — load and downsample to match LOLA 25m pixel grid
from scipy.ndimage import zoom as ndzoom
with rasterio.open(PATHS["shadow"]) as s:
    sh_raw = s.read(1).astype(np.float64)
    sh_nd  = s.nodata
if sh_nd is not None:
    sh_raw = np.where(sh_raw == sh_nd, np.nan, sh_raw)
zf = H / sh_raw.shape[0]
if abs(zf - 1.0) > 0.01:
    sh_ds = ndzoom(np.nan_to_num(sh_raw, nan=0.0), zf, order=1, mode="nearest")
    shadow_matched = sh_ds[:H, :W]
else:
    shadow_matched = sh_raw[:H, :W]

dfsar_cpr, dfsar_tf  = load_band(PATHS["dfsar_cpr"])
dfsar_cov, _         = load_band(PATHS["dfsar_cov"])
minirf_s1,  mrf_tf   = load_band(PATHS["minirf_s1"])
minirf_cov, _        = load_band(PATHS["minirf_cov"])
m3_b1,      m3_tf    = load_band(PATHS["m3"], band=1)
grail_a,    ga_tf    = load_band(PATHS["grail_anom"])
grail_b,    gb_tf    = load_band(PATHS["grail_boug"])
print("   All secondary datasets loaded.")

# ══════════════════════════════════════════════════════════════════════════
# 5. PER-REGION FEATURE EXTRACTION
# ══════════════════════════════════════════════════════════════════════════

print(f"\n[5/6] Extracting features for {n_regions} regions...")
records = []

for i, prop in enumerate(regionprops(region_labels, intensity_image=slope_deg)):
    if i % 1000 == 0:
        print(f"   {i+1}/{n_regions}...", end="\r")
    mask = region_labels == prop.label

    sv = slope_deg[mask];   rv = roughness[mask];  ev = elev[mask]
    shv = shadow_matched[mask]
    shadow_valid = shv[np.isfinite(shv)]

    cy_px, cx_px = prop.centroid
    gx, gy       = rc_xy(transform, cy_px, cx_px)

    records.append({
        # Identity & geometry
        "region_id":           int(prop.label),
        "area_px":             int(prop.area),
        "area_km2":            round(prop.area * res_x * res_y / 1e6, 6),
        "centroid_geo_x":      round(float(gx), 4),
        "centroid_geo_y":      round(float(gy), 4),
        "centroid_row_px":     round(float(cy_px), 2),
        "centroid_col_px":     round(float(cx_px), 2),
        "bbox_row_min":        int(prop.bbox[0]),
        "bbox_col_min":        int(prop.bbox[1]),
        "bbox_row_max":        int(prop.bbox[2]),
        "bbox_col_max":        int(prop.bbox[3]),

        # HIGH-RES terrain (homogeneous by construction)
        "slope_mean_deg":      round(float(np.nanmean(sv)), 4),
        "slope_std_deg":       round(float(np.nanstd(sv)),  4),
        "slope_max_deg":       round(float(np.nanmax(sv)),  4),
        "roughness_mean_m":    round(float(np.nanmean(rv)), 4),
        "roughness_std_m":     round(float(np.nanstd(rv)),  4),
        "elev_mean_m":         round(float(np.nanmean(ev)), 4),
        "shadow_fraction":     round(float(np.nanmean(shadow_valid)), 4)
                               if len(shadow_valid) > 0 else np.nan,

        # MID-RES radar (centroid nearest-pixel)
        "dfsar_cpr_val":       round(geo_sample(dfsar_cpr, dfsar_tf, gx, gy), 6),
        "dfsar_coverage":      round(geo_sample(dfsar_cov, dfsar_tf, gx, gy), 4),
        "minirf_s1_val":       round(geo_sample(minirf_s1,  mrf_tf,  gx, gy), 6),
        "minirf_coverage":     round(geo_sample(minirf_cov, mrf_tf,  gx, gy), 4),

        # COARSE context (regional constant — NOT local discriminators)
        "m3_band1_context":    round(geo_sample(m3_b1,   m3_tf, gx, gy), 4),
        "grail_anom_context":  round(geo_sample(grail_a, ga_tf, gx, gy), 4),
        "grail_boug_context":  round(geo_sample(grail_b, gb_tf, gx, gy), 4),
    })

df = pd.DataFrame(records)
# Round any stray NaN-producing floats
df = df.where(pd.notna(df), other=None)

csv_path = os.path.join(OUT, "candidate_regions.csv")
df.to_csv(csv_path, index=False)

print(f"\n   {len(df)} regions saved → {csv_path}")
print(f"   Slope range: {df['slope_mean_deg'].min():.2f}° – {df['slope_mean_deg'].max():.2f}°")
print(f"   Regions with mean slope < 15°: {(df['slope_mean_deg']<15).sum()} "
      f"({100*(df['slope_mean_deg']<15).mean():.1f}%)")
print(f"   Regions with mean slope < 10°: {(df['slope_mean_deg']<10).sum()} "
      f"({100*(df['slope_mean_deg']<10).mean():.1f}%)")

within_std  = df["slope_std_deg"].mean()
between_std = df["slope_mean_deg"].std()
print(f"\n   Homogeneity check:")
print(f"   Within-region slope std  = {within_std:.3f}°")
print(f"   Between-region slope std = {between_std:.3f}°")
print(f"   Ratio (within/between)   = {within_std/between_std:.3f}")
print(f"   → {'PASS ✓ regions are internally uniform' if within_std < between_std else 'CHECK ⚠'}")

# ══════════════════════════════════════════════════════════════════════════
# 6. VISUALISATION
# ══════════════════════════════════════════════════════════════════════════

print("\n[6/6] Generating candidate_regions_map.png ...")

fig, axes = plt.subplots(1, 3, figsize=(20, 7))
fig.suptitle(
    "Step 2 — Candidate Region Definition\n"
    "Watershed Terrain Segmentation on LOLA Slope  "
    "(Drăguţ & Blaschke 2006; Drăguţ & Eisank 2011)",
    fontsize=12, fontweight="bold", y=1.01
)

ax = axes[0]
im = ax.imshow(slope_deg, cmap="plasma", vmin=0, vmax=30, interpolation="nearest")
plt.colorbar(im, ax=ax, label="Slope (°)", fraction=0.046, pad=0.04)
ax.set_title("A — LOLA Slope (°)\n(segmentation input)", fontsize=10)

ax = axes[1]
smean_img = np.full((H, W), np.nan)
for row in records:
    smean_img[region_labels == row["region_id"]] = row["slope_mean_deg"]
im2 = ax.imshow(smean_img, cmap="RdYlGn_r", vmin=0, vmax=25, interpolation="nearest")
plt.colorbar(im2, ax=ax, label="Region mean slope (°)", fraction=0.046, pad=0.04)
ax.set_title(f"B — {n_regions} Terrain-Homogeneous Regions\n(coloured by mean slope)", fontsize=10)

ax = axes[2]
acc_img = np.full((H, W), np.nan)
for row in records:
    s  = row["slope_mean_deg"] or 0
    r  = row["roughness_mean_m"] or 0
    sf = row["shadow_fraction"] if row["shadow_fraction"] else 0.5
    acc = max(0.0, 1 - s/30) * max(0.0, 1 - r/50) * (1 - 0.3*sf)
    acc_img[region_labels == row["region_id"]] = acc
im3 = ax.imshow(acc_img, cmap="RdYlGn", vmin=0, vmax=1, interpolation="nearest")
plt.colorbar(im3, ax=ax, label="Accessibility proxy (0→1)", fraction=0.046, pad=0.04)
ax.contour(slope_deg, levels=[15], colors=["red"], linewidths=0.7,
           linestyles="--", alpha=0.8)
ax.legend(handles=[mpatches.Patch(color="green", label="Safe (<15°)"),
                   mpatches.Patch(color="red",   label="Risky (>15°)")],
          loc="lower right", fontsize=8)
ax.set_title("C — Accessibility Preview\n(red dashed = 15° rover slope limit)", fontsize=10)

plt.tight_layout()
map_path = os.path.join(OUT, "candidate_regions_map.png")
plt.savefig(map_path, dpi=150, bbox_inches="tight")
plt.close()
print(f"   Saved: {map_path}")

print("\n" + "=" * 65)
print("STEP 2 COMPLETE")
print(f"  CSV → {csv_path}")
print(f"  MAP → {map_path}")
print("=" * 65)
```
